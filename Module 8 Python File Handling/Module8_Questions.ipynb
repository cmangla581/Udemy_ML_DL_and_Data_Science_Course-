Q1. Write a function that reads the contents of a file named `sample.txt` and prints each line.

```python
# The code for the above question is as: 

def read_file(filename): 
    with open(filename, 'r') as file: 
        for line in file: 
            print(line.strip())  

# Teasting will be done as 

read_file('example.txt') 

# The content in the file gets printed 
```

Q2. Write a function that writes a list of strings to a file named `output.txt`, with each string on a new line.

```python
# The code for the above question is as: 

def write_file(lines, filename): 
    with open(filename, 'w') as file: 
        for line in lines: 
            file.write(line + '\n') 

# Now, we will be writing the text file as:
write_file(['Hello', 'World'], 'output.txt') 
```

Q3.  Write a function that copies the contents of a file named `source.txt` to a new file named `destination.txt`.

```python
# Te code for the above question is as: 

def copy_file(source, destination): 
    with open(source, 'r') as src: 
        with open(destination, 'w') as dest: 
            dest.write(src.read())  

copy_file('source.txt', 'destination.txt') 
```

Q5. Write a function that reads the contents of a file named `document.txt` and returns the number of words in the file. 

```python
# The code for the above the Question   
def count_words(filename): 
    with open(filename, 'r') as file:  
        text = file.read() 
        words = text.split() 
        return len(words)  

# Printing the len of the words 
print(count_words('example.txt')) 

# There area total of around 7 words in the whole document 
```

Q6. Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.

```python
# The code for the above question for the given ord and in the file named as data.txt as: 

def find_and_replace(filename, old_word, new_word): 
    with open(filename, 'r') as file: 
        text = file.read() 
    new_text = text.replace(old_word, new_word) 

    with open(filename, 'w') as file: 
        file.write(new_text) 


find_and_replace('example.txt', 'new', 'old') 
```

Q7. Write a function that reads the contents of a file named `reverse.txt` and prints each line in reverse order.

```python
# The function that reads the contents of a file as reverse.txt and prints each line in reverse order 

def read_reverse(filename): 
    with open(filename, 'r') as file: 
        lines  = file.readlines() 
    for line in reversed(lines): 
        print(line.strip()) 

# Reading the text as: 
read_reverse('example.txt')   

# The lines are writing in reversed 
```

Q8. Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.

```python
# The code for  the above question takes place as: 

# Counting lines, words and characters 
def count_lwc(filename): 
    with open(filename, 'r') as file: 
        lines = file.readlines() 
        words = sum(len(line.split()) for line in lines)   
        characters =  sum(len(line) for line in lines)
    return len(lines), words, characters  

# Printing the character as: 
count_lwc('example.txt')  

# It returns the number of lines, words and the characters 
```

Q9.  Write a function that merges the contents of multiple files into a single file named `merged.txt`.

```python
# The code for the above question is as: 

def merge_files(file_list, output_file): 
    with open(output_file, 'w') as outfile: 
        for fname in file_list: 
            with open(fname, 'r') as infile:
                outfile.write(infile.read() + '\n') 

merge_files(['example.txt', 'examples.txt'], 'merged.txt')  
```

Q10.  Write a function that splits a large file named `large.txt` into smaller files of 100 lines each.

```python
def split_files(filename, lines_per_file): 
    with open(filename, 'r') as file: 
        lines = file.readlines() 
    for i in range(0, len(lines), lines_per_file): 
        with open(f'{filename}_part{i//lines_per_file + 1}.txt', 'w') as part_file: 
            part_file.writelines(lines[i:i + lines_per_file]) 

split_files('example.txt', 100) 
```

Q11. Write a function that creates a log file named `activity.log` and writes log messages with timestamps.

```python
# The code for the above question is as:  

import datetime 

def log_message(message, filename = 'activity.log'): 
    timestamp = datetime.datetime.now().isoformat()  
    with open(filename, 'a') as file:
        file.write(f'[{timestamp}] {message}\n') 

log_message('This is a log message') 
```

Q12. Write a function that reads a binary file named `image.bin` and writes its contents to another binary file named `copy_image.bin`.

```python
# The code for the above question is as: 

def copy_binary_file(source, destination): 
    with open(source, 'rb') as src:  
        with open(destination, 'wb') as dest: 
            dest.write(src.read()) 

copy_binary_file('example.bin', 'copy_example.bin')  
```

Q13. Write a function that reads a CSV file named `data.csv` and prints its contents as a list of dictionaries.

```python
# The code for the above question is as: 

import csv 
def read_csv_as_dicts(filename):
    with open(filename, 'r') as file:
        reader = csv.DictReader(file) 
        return list(reader) 

#For testing, we will be applying as: print(read_csv_as_dicts('data.csv')) 
```

Q14. Write a function that reads a JSON file named `data.json` and prints its contents as a Python dictionary. 

```python
# The function that reads a json file and makes the input as: 

import json 

def read_json(filename): 
    with open(filename, 'r') as file:  
        data = json.load(file) 
        return data 
```

Q15.  Write a function that attempts to read a file named `protected.txt` and handles any permission errors gracefully by printing an error message.

```python
# The code for the above question is: 

def read_protected_file(filename): 
    try: 
        with open(filename, 'r') as file: 
            print(file.read()) 

    except PermissionError as e: 
        print(f'Permission Error: {e}') 

# Now, we will be carrying out the testing as: 
# read_protected_file('protected.txt) 
```
