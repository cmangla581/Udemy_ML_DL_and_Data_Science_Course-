                                                         File Handling

File handling is a crucial part of the any of the programming languages. Python provides the built in functions and methods to read and write from the files, both of the text and binary.  This lesson covers the basic part of the handling which includes the reading, writing, text files and binary. 

```python
#Example of the file reading 

with open('example.txt', 'r') as file: 
    content = file.read() 
    print(content) 

# The contents of the file get printed 
```

```python
# Reading a file line by line 

with open('example.txt', 'r') as file: 
    for line in file: 
        print(line) 

# Here, as we can see that the information get printed line by line. 
```

```python
# Now, suppose we want to remove the new line characters between the lines, hence it will print the lines as such 

with open('example.txt', 'r') as file: 
    for line in file: 
        print(line.strip())   

```

Writing a file: Now we will write a file and will do the overwriting. 

```python
with open('example.txt', 'w') as file: 
    file.write("Hello World\n") 
    file.write("This is a new line")
```

In this code, we can see that the old lines get removed and the new lines get printed. Hence this  an example of overwriting. 

```python
# Writing the file  without overwriting 
with open('example.txt', 'a') as file: 
    file.write("\nAppend Operation taking place")
```

Here, as we can see that the lines get appended in the file without the deletion of the previos file. 

Now, we will write the code of writing the list of lines to the file in Python  

```python
lines = ['first line \n', 'Second Line \n', 'Third Line\n'] 

with open('examples.txt', 'a') as file: 
    file.writelines(lines) 
```

```python
# Binary Files  

data = b'\x00\x01\x02\x03\x04' 

with open('example.bin', 'wb') as file: 
    file.write(data) 
```

In the above code, we can see that the binary files get printed successfully 

```python
with open('example.bin', 'rb') as file: 
    content = file.read()
    print(content) 

# The contents get printed 
```

```python
# read the content from a sourcce text file and write to a destination text file. 
with open('source.txt', 'r') as source_file: 
    content = source_file.read() 

with open('destination.txt', 'w') as destination_file: 
    destination_file.write(content) 
```

Q. Read the Text File and count the number of the lines, words and the characetrs. 

```python
# Counting lines, words, nd the characters in a text file 

def count_text_file(file_path): 
    with open(file_path, 'r') as file: 
        lines = file.readlines()
        line_count = len(lines) 
        word_count = sum(len(line.split()) for line in lines) 
        char_count = sum(len(line) for line in lines) 
    return line_count, word_count, char_count  

file_path = 'example.txt' 
lines, words, characters = count_text_file(file_path)  

print(f'Lines : {lines}, Words : {words}, Characters: {characters}') 
```

The w+ mode in Python is used to open a file for both reading and writing. If the file does not exist, it will be created. If the file exists, its content will be overwritten
