# Inheritance 

Inheritance is basically an important concept in Python OOP which basically allows us to inherit the  attributes and methods from another class.  This lesson mainly covers the single inheritance and multiple inheritance and also demonstrating, on how to create and use them in Python. 

```python
# Example on Inheritance in Pythn OOP 

class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows 
        self.doors = doors 
        self.enginetype = enginetype 

    def drive(self):
        print(f"The person will drive the {self.enginetype} car")  

car1 = Car(4,5, "petrol")  
print(car1.windows)
print(car1.doors)
print(car1.enginetype)
car1.drive() 
```

The above example is as simple as we do for the normal Object Oriented programming. Now, we try the the concept of inheritance in it by making a different subclass of Tesla, as its also a car and alon with it, it would be containing the attributes of the parent class Car, along with new features also. 

Hence, basically, we can say that in inheritance, a class can inherit the attributes from another class, making the latter one as the parent class. 

```python
class Tesla(Car):
    def __init__(self, windows, doors, enginetype, is_selfdriving):
        super().__init__(windows, doors, enginetype) 
        self.is_selfdriving = is_selfdriving  

    # making an instance method working takes place as: 
    def selfdriving(self):
        print(f"Tesla supports self driving : {self.is_selfdriving}") 

# Now, we will try the working of the above code as: 
tesla1 = Tesla(4,5, "electric", True) 
print(tesla1.windows)
print(tesla1.doors)
print(tesla1.enginetype)
print(tesla1.is_selfdriving) 
tesla1.selfdriving() 
```

```python
# Also since, in the Car parent Class, we dave the drive function as the self instance function 
tesla1.drive() 
```

# Multiple Inheritance  

Multiple Inheritance is the basic concept of object oriented programming where a class inherits from more than one base class. 

```python
# base Class 1  

class Animal: 
    def __init__(self, name): 
        self.name = name 

    def speak(self): 
        print("Subclass must implement this method") 

# Blase Class 2 
class Pet: 
    def __init__(self,owner): 
        self.owner = owner 

# Now, we will be making the derived class in the code to show the working as follows: 
class Dog(Animal, Pet): 
    def __init__(self, name, owner): 
        Animal.__init__(self, name) 
        Pet.__init__(self, owner) 

    def speak(self): 
        return f"{self.name} say woof"  

# creating the object 
dog = Dog("Buddy", "Krish") 
print(dog.name) 
print(dog.owner) 

print(dog.speak())  
print(f"Owner:{dog.owner}") 
```
