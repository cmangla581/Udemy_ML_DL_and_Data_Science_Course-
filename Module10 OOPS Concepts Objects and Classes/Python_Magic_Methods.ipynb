# Magic Methods 

Magic Methods in Python are also know as the dunder methods which are basically the specual methods that basically start and end with double underscores. These methods enable us to define the behaviour of the objects for the built in operations such a the built in operations, comparisons and more. 

Magic methods ae basically like the pre defined methods in Python that you can use to change the behaviour of your objects. Some of the common magic methods include:  

```python
'''

__init__ : It initializes a new instance of class  
__str__ : It returns a string representation of an object 
__repr__ : It returns an official string representation of the object 
__len__ : It returns the length of an object 
__getitem__ : gets an item from container 
__setitem__ : It sets an item in the container 

''' 
```

We can perform the overriding function using the string which can be easily done in the following ay as: 

```python
class Person: 
    def __init__(self,name, age): 
        self.name = name 
        self.age = age 
        

person = Person("Krish", 34) 
print(person)  

# The location of thr object gets printed here 
```

```python
class Person: 
    def __init__(self, name, age): 
        self.name = name 
        self.age = age 

    def __str__(self): 
        return f"{self.name}, {self.age} years old" 
    
person = Person("Krsih", 34) 
print(person)  

# Here due to the string method the result gets printed 
```

```python
# Now. lets try the smae above example with __repr__ functionality f Python as: 

class Person: 
    def __init__(self, name, age): 
        self.name = name 
        self.age = age 

    def __str__(self): 
        return f"{self.name}, {self.age} years old" 
    
    def __repr__(self): 
        return f"Person(name= {self.name}, age= {self.age})" 
    
person = Person("Krsih", 34) 
print(person)  

print(repr(person)) 

# Here, as we can see that these 2 functions get printed 
```
