# Instance Variable and Methods 

```python
class Dog: 
    # init is basically a constructor which helps in defining all the attributes handly 
    def __init__(self, name, age):
        self.name = name 
        self.age = age  

dog1 = Dog("Buddy", 3) 
print(dog1) 
print(dog1.name) 
print(dog1.age)  

dog2 = Dog("Lucy", 7) 
print(dog2) 
print(dog2.name) 
print(dog2.age) 
```

In thw above code we can see that the object has the  attributes written at once and hence we can see that the name and age get printed at once by calling the constructor. 

```python
# Defining a class with instance methods 

class Dog: 
    def __init__(self, name, age): 
        self.name = name 
        self.age = age 

    # Now, lets create an instance method related to dog 
    def bark(self): 
        print(f"{self.name} says woof") 


dog1 = Dog("Buddy", 3) 
dog1.bark() 

dog2 = Dog("Lucy", 5) 
dog2.bark()  

# Here, according to the names, the different names have been printed 
```

Now, by using the class and objects, we will be making the modelling of a Bank Account to show all the processing in a bank. 

```python
class BankAccount: 
    def __init__(self, owner, balance = 0): 
        self.owner = owner 
        self.balance = balance

    def deposit(self, amount): 
        self.balance += amount 
        print(f"{amount} is deposited, New balance is {self.balance}") 

    def withdraw(self, amount): 
        if amount > self.balance: 
            print("Insufficient Funds") 

        else: 
            self.balance -= amount 
            print(f"{amount} is withdrawwn. New balance is {self.balance}") 

    def  get_balance(self): 
        return self.balance 
```

```python
# Now, here we will be executing the above code and see the working of the code as 

account = BankAccount("Krish", 5000) 
print(account.balance)  

# The nalance of the account gets prinnted 
```

```python
# Now, we will be calling the instance methods as following: 
account.deposit(100) 
```

```python
account.withdraw(300) 
```

```python
# Now, we have to get the final balance of the whole account 
print(account.get_balance())
```
