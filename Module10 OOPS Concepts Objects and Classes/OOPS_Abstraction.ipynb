# Abstraction  
Abstraction is the concept of hiding the complex implementation details and showing oonly the necessary features of an object. This helps in reducing the programming complexity and effort. 

Like, here we can take the example of a washing machine. In a washing machine, on just clicking the buttons, we can easily operate all the features and we don't need to now the inside mechanism of the machine. hence, this is why we use the abstraction method. 

```python
# Abstract base Classes work as: 
from abc import ABC, abstractmethod

class Vehicle(ABC):
    def drive(self): 
        print("The vehicle is self driving") 

    @abstractmethod 
    def start_engine(self): 
        pass 

class Car(Vehicle): 
    def start_engine(self): 
        print("Car engine started")  

def operate_vehicle(vehicle): 
    vehicle.start_engine()  

car = Car() 
operate_vehicle(car) 
```

# Conclusion   
Hence the differences between the 4 pillars of OOPS include the:  

1. Encapsulation : It basically means the wrapping of data together and restricting direct access. Like a capsule where medicine is inside and is protected.  

2. Inheritance : It isbasically that one class can inherit properties and methods from another class. It is like a child inheriting the features from the parents. 

3. Polymorphism : Same function name but diferent behaviour depending on the object. Like + works on the numbers for addition and the strings are there for concatenation. 

4. Abstraction : Showing only the important details and hiding the complex implemantation. Like driving a car without knowing, how the engine works. 

We can take even simpler examples as: 

1. Encapsulation : car parts are inside one system 
2. Inheritance : Electric car and Sports car can inherit from car 
3. Polymorphism : start() may work differently for different cars 
4. Abstraction : You use only steering, brake and accelerator and not the internal engine system. 

In one line we can say that : 
1. Encapsulation : protects data 
2. Inheritance : reuse code 
3. Polymorphism : same interface and same behaviour 
4. Abstraction : hide complexity
