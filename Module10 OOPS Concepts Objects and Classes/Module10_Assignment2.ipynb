Q1. Create a base class named `Animal` with attributes `name` and `species`. Create a derived class named `Dog` that inherits from `Animal` and adds an attribute `breed`. Create an object of the `Dog` class and print its attributes. 

```python
# the code for the above question as: 

class Animal: 
    def __init__(self, name, species): 
        self.name = name 
        self.species = species 

# Creating a derived class 
class Dog(Animal): 
    def __init__(self, name,species, breed): 
        super().__init__(name, species) 
        self.breed = breed  

# Testing of the code as: 
dog = Dog("Buddy", "Canine", "Golden Retriever") 
print(dog.name, dog.species, dog.breed) 
```

Q2. In the `Dog` class, override the `__str__` method to return a string representation of the object. Create an object of the class and print it. 

```python
# the code for the above question is as: 

class Dog(Animal): 
    def __init__(self, name, species, breed):
        super().__init__(name, species)
        self.breed = breed  

    # Making the string method 
    def __str__(self): 
        return f"Dog(Name: {self.name}, Species: {self.species}, Breed: {self.breed})" 
    
# The testing of the code can be done as: 
dog = Dog('Buddy', 'Canine', 'Golden Retriever') 
print(dog) 
```

Q5.   In the `Athlete` class, override the `walk` method to print a different message. Create an object of the class and call the `walk` method. Use the `super()` function to call the `walk` method of the `Walker` class. 

```python
# The code for the above question works as: 

class Walker: 
    def walk(self): 
        print("Walking---") 

class Runner: 
    def run(self):
        print("Running-----") 

class Athlete(Walker, Runner): 
    # overriding the method as 
    def walk(self): 
        print("Athlete Walking-----")  
        super().walk() 
        

athlete = Athlete() 
athlete.walk() 
```

Q7.  Create a class named `A` with a method `show` that prints a message. Create two derived classes `B` and `C` that inherit from `A` and override the `show` method. Create a class `D` that inherits from both `B` and `C`. Create an object of the `D` class and call the `show` method. Observe the method resolution order. Its also known as Damond Problem 

```python
# The code for the above question is as: 

class A: 
    def show(self):
        print("A's show method") 

class B(A): 
    def show(self): 
        print("B's show method") 

class C(A): 
    def show(self): 
        print("C's show method") 

class D(B,C):
    pass 

d = D() 
d.show()   

'''
In the above code, it prints B's show method as because B is before C and hence, it prints the result of B 
''' 
```

Q9.  Create a class named `Person` with an attribute `name`. Create a class named `Employee` with an attribute `employee_id`. Create a derived class `Manager` that inherits from both `Person` and `Employee`. Use the `super()` function to initialize the attributes. Create an object of the `Manager` class and print its attributes. 

```python
# The code for the above question as: 

class Person: 
    def __init__(self, name): 
        self.name = name 

class Employee: 
    def __init__(self, employee_id): 
        self.employee_id = employee_id 

class Manager(Person, Employee): 
    def __init__(self, name, employee_id): 
        Person.__init__(self, name) 
        Employee.__init__(self, employee_id) 

# The testing of the code as: 
manager = Manager("John", "1234") 
print(manager.name)
print(manager.employee_id)
```

Q10. Create a class named `Vehicle` with a method `start` that prints a starting message. Create a derived class `Car` that overrides the `start` method to print a different message. Use the `super()` function to call the `start` method of the `Vehicle` class. Create an object of the `Car` class and call the `start` method. 

```python
# The code for the above question is as: 

class Vehicle: 
    def start(self): 
        print("Vehicle is starting-----") 

class Car(Vehicle): 
    def start(self): 
        print("Car starting") 
        super().start() 

# The testing of the code 
car = Car() 
car.start() 
```

Q12. Create a class named `Base1` with an attribute `a`. Create a class named `Base2` with an attribute `b`. Create a class named `Derived` that inherits from both `Base1` and `Base2` and adds an attribute `c`. Initialize all attributes using the `super()` function. Create an object of the `Derived` class and print its attributes. 

```python
# The code for the above question is as: 

class Base1: 
    def __init__(self, a): 
        self.a = a 

class Base2: 
    def __init__(self, b): 
        self.b = b 

class Derived(Base1, Base2): 
    def __init__(self, a,b,c): 
        Base1.__init__(self,a) 
        Base2.__init__(self,b)  
        self.c = c 

# The testing of the code as: 
derived = Derived(1,2,3) 
print(derived.a, derived.b, derived.c) 
```

Q14.  Create a base class named `Bird` with a method `speak`. Create two derived classes `Parrot` and `Penguin` that override the `speak` method. Create a list of `Bird` objects and call the `speak` method on each object to demonstrate polymorphism.

```python
# The code for the above question is as:  
class Bird: 
    def speak(self): 
        pass 

class Parrot(Bird): 
    def speak(self): 
        print("Parrot saya: Squak") 

class Penguin(Bird): 
    def speak(self): 
        print("Penguin says: Honk")  

def bird_speak(bird): 
    bird.speak() 

parrot = Parrot() 
penguin = Penguin() 
parrot.speak()
penguin.speak() 

bird_speak(parrot) 
bird_speak(penguin) 
```

Q15.  Create a base class named `Device` with an attribute `brand`. Create a derived class `Phone` that inherits from `Device` and adds an attribute `model`. Create another base class `Camera` with an attribute `resolution`. Create a derived class `Smartphone` that inherits from both `Phone` and `Camera`. Create an object of the `Smartphone` class and print its attributes. 

```python
# The code for the above question is as: 

class Device: 
    def __init__(self, brand): 
        self.brand = brand 

class Phone(Device): 
    def __init__(self, brand, model): 
        super().__init__(brand)
        self.model = model 

class Camera: 
    def __init__(self, resolution): 
        self.resolution = resolution 

class Smartphone(Phone, Camera): 
    def __init__(self, brand, model, resolution): 
        Phone.__init__(self, brand, model) 
        Camera.__init__(self, resolution) 

# Testing of the code as: 
smartphone = Smartphone("Apple", "Iphone12", "12MP") 
print(smartphone.brand, smartphone.model, smartphone.resolution) 

    
```
