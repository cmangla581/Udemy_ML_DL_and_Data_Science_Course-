# Polymorphism  

Polymorphism is a  core concept in Object Oriented Programming (OOP) that allows objects of different classes  to be treated as objects of a common super class. It provides a way to perform a singe action in the different forms. Polymorphism is typically achieved through method overriding and interfaces.  

Plymorphism basically means multiple forms or many forms as its a combinations of the words multi and forms. 

# Method Overriding 
Method overriding basically allows a child class to provide a specific implementation of a method that is already defined in its parent class. 

```python
# The example of the code can be defined as: 

class Animal: 
    def speak(self): 
        return "Sound of the animal"  

# Derived Class 1 
class Dog(Animal): 
    def speak(self): 
        return "Woof!" 

# Derived Class 2 
class Cat(Animal): 
    def speak(self):
        return "Meow!"  

# Here, we can add one more function that demonstrates polymorphism 
def animal_speak(animal): 
    print(animal.speak()) 

dog = Dog() 
cat = Cat() 
print(dog.speak()) 
print(cat.speak())  

animal_speak(dog)  
# Here, as we can see that the polymorphism is denoted in this class as because as the objects of 
```

```python
# Polymorphism with Functions and Methods   
# The polymorphism with objects andmethods basically mean that the objects of different types can be processed in same  way 

class Shape: 
    def area(self): 
        return "The area of the figure"  

class Rectangle(Shape): 
    def __init__(self, width, height): 
        self.width = width 
        self.height = height  

    def area(self):  
        return self.width * self.height  

# Derived Class 2 
class Circle(Shape): 
    def __init__(self, radius): 
        self.radius = radius 

    def area(self): 
        return 3.14*self.radius** 2  

# Functions that demonstrates polymorphism 

def print_area(shape): 
    print(f"The area is {shape.area()}") 

rectangle = Rectangle(4,5) 
circle = Circle(3) 

print_area(rectangle) 
print_area(circle)  

# Hence the area of the rectangle and the circles have been printed 
```

# Polymorphism with Abstract Base Classes 

Abstract Base Classes are used to define the common methods for a group of related objects. They can enforce the derived classes, implement the particular methods, promote consistency across different implementations. 

```python
# Example form a code as: 
from abc import ABC, abstractmethod 

# Defining an abstract class:
class Vehicle(ABC): 
    @abstractmethod 
    def start_engine(self):
        pass 

# Derived Clss 1 
class Car(Vehicle): 
    def start_engine(self): 
        return "Car Engine started" 
    
# Derived Class 2 
class Motorcycle(Vehicle): 
    def start_engine(self): 
        return "Motorcycle Engine started"
    
def start_vehicle(vehicle): 
    print(vehicle.start_engine()) 

# Create objects of Car and Motorcycle 
car = Car() 
motorcycle = Motorcycle() 

start_vehicle(car) 
start_vehicle(motorcycle) 
```
