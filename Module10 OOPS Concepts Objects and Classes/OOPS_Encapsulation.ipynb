Encapsulation and Abstraction are the 2 fundamental principles of Object Oriented Programming that help in the designing the robot , maintainable and reusable code. Encapsulation involves the building data and methods that operate on the data within a  single unit, while the abstraction involves the hiding of the complex implementation details and exposing only the necessary features. 

## Encapsulation  
Encapsulation is the concept of wrapping the data (variables) and methods, together as a single unit. It restricts the direct access to some of the object's components,which is a means of preventing the accidental interference and misuse of the data. 

```python
## Encapsulation with the getter and setter method  
# Public, private and protected variables.  

class Person: 
    def __init__(self, name, age): 
        self.name = name  # Public variables 
        self.age = age    # Public variables 

person = Person("Krish", 28) 
print(person.name) 

# Here, the name of the person gets printed according to the encapsulation method 
```

# Encapsulation with Getter and Setter 


```python
class Person: 
    def __init__(self, name, age): 
        self.__name = name  # Private variable 
        self.__age = age   # private variable   

    # getter method for name 
    def get_name(self): 
        return self.__name  
    
    # setter method for name 
    def set_name(self, name):
        self.__name = name 

    # Getter method for age 
    def get_age(self): 
        return self.__age
    
    def set_age(self, age): 
        if age > 0:
            self.__age = age 

        else:  
            print("Age cannot be negative") 

person = Person("Krish", 34) 
# Access and modify the private variables from getter and setter  

print(person.get_name()) 
print(person.get_age())  

person.set_age(35)  
print(person.get_age()) 

person.set_age(35) 

person.set_age(-1)  

# here as we can see that the results have got printed correctly 
```

So, the basic difference is that the private vriables cannot be accessed outside the class and the protected can be accessed from a derived class.
