# Generators 

Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means that they generate values on the fly and do not store them in memory

```python
def square(n):  # Here, we are giving the value of the n as 3 
    for i in range(3): 
        return i**2 
```

```python
square(3)  
# here it gives 0 as it iterates through all the 
```

```python
def square(n): 
    for i in range(3):
        yield i**2 
```

```python
square(3)  
# Here the square of 3 is a type of generator. 
```

```python
# Code for calling the iterator in genertor takes place as: 
for i in square(3): 
    print(i) 
```

```python
# Suppsoe we store the square 3 in an object as: 
a = square(3) 
a 
```

```python
# now we use the next operator as: 
next(a)  

# here, after printing 4, next of a starts giving error 
```

```python
def my_generator():
    yield 1 
    yield 2 
    yield 3 

gen = my_generator() 
gen  
```

```python
# Now we apply the nect method which takes place as: 
next(gen) 

# Here it gives 3 at the end, just after this it will give the error 
```

```python
for val in gen: 
    print(val) 
```

# Practical examples of Generators  
Generators are particularly useful in for reading the large files because they allow you to process one line at a time without leading the  entire file memory. 

```python
def read_large_file(file_path): 
    with open(file_path, 'r') as file: 
        for line in file:
            yield line
```

```python
file_path = 'large_file.txt' 

for line in read_large_file(file_path): 
    print(line.strip()) 
```

Basically, the main difference include that the generators use the yield keyword and the iterators use the iter keyword and generators save the local keyword.
