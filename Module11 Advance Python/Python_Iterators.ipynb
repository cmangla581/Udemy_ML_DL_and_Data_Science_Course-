# Iterators  
Iterators are the advance Python concepts that allow for the efficient looping and memory management. Iterators provide a way to access the elements of a collection sequentially without exposing the  underlying structure. 

```python
# example through the code 
my_list = [1,2,3,4,5,6] 
for i in my_list: 
    print(i)  

# The numbers in the list have been printed 
```

```python
type(my_list) 
# here it shows the type   
```

```python
print(my_list) 
# Here all the elements in the list are getting printed 
```

```python
# Iterator 
iterator = iter(my_list) 
print(type(iterator))  

print(iterator) 
```

Here, in the abve code, as we can see that the iterator does not give the elements of the list. Here, it follows the lazy loading method. 

For the showing of the elements, we have to use the "next iterator" function. 

```python
next(iterator) 
```

In the above code, as we can see that as we keep running the code, we keep getting the next numbers and at the end, the iterator prints the error and hecne this is how it works. 

```python
next(iterator) 
# Error generated 
```

Coding it properly with the try and except as: 

```python
iterator = iter(my_list)  
```

```python
try: 
    print(next(iterator))
except StopIteration: 
    print("There are no elements in the iterator") 
```

```python
# String Iterators 
my_string = "Hello"  
string_iterator =  iter(my_string) 

print(next(string_iterator)) 
print(next(string_iterator))  

# Here the first 3 letters are getting printed 
```
