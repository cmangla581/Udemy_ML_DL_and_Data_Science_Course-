# Decorators  
Decorators are a powerful and flexible features in Python that allows you to modify the behaviour of a function or class method. They are commonly used to add the functionality too the functions or methods withut modifying their actual codes. This lesson covers the basics of the decoorators, including on how to ceate and use them.  

```python
# function copy '
def welcome(): 
    return "Welcome to the advance python course"  
welcome()  

# The code gets printed here perfectly 
```

```python
# Try deleting the function here as: 
del welcome  
print(wel()) 
```

Use of Decorators 

```python
def main_welcome(func): 

    def sub_welcome_method(): 
        print("Welcome to the advance python course") 
        func() 
        print("Please learn these concepts properly")
    return sub_welcome_method() 
```

```python
def course_introduction(): 
    print("This is an advanced python course") 

course_introduction() 
```

```python
# Now, we call the course introduction function inside the main_welcome function and it will take place as: 

main_welcome(course_introduction) 

# Here as we can see that the function get printed here easily and hence it takes place like this. 
```

```python
# Here we can use a decorator keyword to write the main welcome function and hence we will see that: 

@main_welcome 
def course_function(): 
    print("This is an advaance python function") 

# As, we can see that all the lines are getting printed here 
```

```python
# The code for the function is as follows: 
def my_decorator(func): 
    def wrapper(): 
        print("Something big is going to happen") 
        func() 
        print("Something is happening after the function iis called") 
    return wrapper()
```

```python
@my_decorator  
def say_hello():
    print("Hello!") 
```

```python
# Function with the keyword and positional arguments 
def repeat(n): 
    def decorator(func): 
        def wrapper(*args, **kwargs): 
            for _ in range(n): 
                func(*args, **kwargs)
        return wrapper

    return decorator      
```

```python
@repeat(3) 
def say_hello(): 
    print("Hello")
```

```python
say_hello() 

# Hello gets printed 3 times 
```

# Conclusion 
Decorators are a powerful tool in Python for extending and modifying the behaviour of functions and methods. They provide a clean and readable way to add functionality in the code. such as logging, timing access and control and more without changing the original code.
