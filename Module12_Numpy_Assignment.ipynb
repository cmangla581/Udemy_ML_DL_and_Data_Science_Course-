Q2. Create a NumPy array of shape (3, 3) with values from 1 to 9. Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1). 

```python
# The code for the abbove question is as: 
import numpy as np 
array = np.arange(1,10).reshape((3,3)) 
print("Original Array") 
print(array) 

# Normalizing the array can be doenn as:  
mean = np.mean(array) 
std_dev = np.std(array) 

normalized_array = (array - mean)/ std_dev 

print("Normalized Array") 
print(normalized_array) 
```

Q1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues. 

```python
import numpy as np 
matrix = np.random.randint(1,11, size = (3,3)) 
print("Original Matrix") 
print(matrix) 

# Computing the determinant 
determinant = np.linalg.det(matrix) 
print("Determinant value: ", determinant) 

# Computing the inverse 
inverse = np.linalg.inv(matrix) 
print("Inverse: ") 
print(inverse) 

# Eigen values can be calculated as: 
eigenvalues = np.linalg.eigvals(matrix) 
print("Eigen values: ", eigenvalues)
```

Q1. Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements. 

```python
# The code fr the question is as: 
import numpy.ma as ma 

array = np.random.randint(1,21, size = (4,4)) 
masked_array = ma.masked_array(array, 10) 
print(masked_array) 

sum_masked = masked_array.sum() 
print("Sum of masked elements: ", sum_masked) 
```

Q2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements. 

```python
# The cde for  the question is as: 
array = np.random.randint(1,21, size = (3,3)) 
masked_array = ma.masked_array(array, mask = np.eye(3, dtype = bool)) 

print("Original Array") 
print(array) 

print(masked_array) 

mean_unmasked = masked_array.mean() 
masked_array = masked_array.filled(mean_unmasked) 
print("Modified masked array") 
print(masked_array) 
```
