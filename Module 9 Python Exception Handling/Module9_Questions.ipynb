# Exception Handling with try, except, and finally 

Q1. Write a function that takes two integers as input and returns their division. Use try, except, and finally blocks to handle division by zero and print an appropriate message.

```python
# The code for the following code is: 

def divide(a,b): 
    try: 
        result = a/b 
    except ZeroDivisionError as e: 
        print(f'Error:{e}') 
        result = None 
    finally: 
        print("Execuion Complete") 
    return result  

print(divide(10,5)) 
print(divide(5,0)) 
```

Q2.  Write a function that reads the contents of a file named `data.txt`. Use try, except, and finally blocks to handle file not found errors and ensure the file is properly closed.

```python
# The code for the above question which take lace as 

def read_file(filename): 
    try: 
        file = open(filename, 'r') 
        content = file.read() 
        return content 
    
    except FileNotFoundError as e: 
        print(f"Error:{e}") 

    finally:
        try:
            file.close()
        except NameError:
            pass 

print(read_file('example.txt')) 
```

Q3. Write a function that takes a list of integers and returns their sum. Use try, except, and finally blocks to handle TypeError if a non-integer value is encountered and print an appropriate message.

```python
# The code for the above question is as: 
def sum_list(lst): 
    total = 0 
    try: 
        for item in lst: 
            total += item 
    except TypeError as e:  
        print(f"Type Error:{e}") 
        total = None 
    finally:  
        print("Execution Complete") 
    return total  

print(sum_list([1,2,3,4,5])) 
```

Q6. Write a function that performs nested exception handling. It should first attempt to convert a string to an integer, and then attempt to divide by that integer. Use nested try, except, and finally blocks to handle ValueError and ZeroDivisionError and print appropriate messages.

```python
# The code for the above question is as: 
def nested_exception_handling(s): 
    try: 
        try: 
            num = int(s)  
        except ValueError as e:  
            print(f"Conversion error:{e}") 
            num = None 
        finally: 
            print("Conversion about to complete") 
        if num is not None: 
            try: 
                result = 10/num  
            except ZeroDivisionError as e: 
                print(f"Division Error:{e}")  
                result = None  
            finally: 
                print("Division attempt complete") 
            return result 
    finally: 
        print("Overall execution complete") 

print(nested_exception_handling('0'))  # None
print(nested_exception_handling('a'))  # None
print(nested_exception_handling('2'))  # 5.0 
```

Q8. Write a function that attempts to open a URL and read its contents. Use try, except, and finally blocks to handle network-related errors and print an appropriate message.

```python
# The code for the above question is as: 

import requests 
def read_url(url):
    try: 
        response = requests.get(url) 
        response.raise_for_status 
        return response.text 
    except requests.RequestException as e:
        print(f"Network Error:{e}") 
        return None 
    finally: 
        print("execution Complete") 


print(read_url('https://jsonplaceholder.typicode.com/posts/1'))
print(read_url('https://nonexistent.url'))


```

Q10. Define a custom exception named `NegativeNumberError`. Write a function that raises this exception if a negative number is encountered in a list. Use try, except, and finally blocks to handle the custom exception and print an appropriate message.

```python
class NegativeNumberError(Exception): 
    pass 

def check_for_negativs(lst): 
    try: 
        for num in lst: 
            if num < 0: 
                raise NegativeNumberError(f"Negative number fund:{num}") 
            
    except NegativeNumberError as e: 
        print(f"Error:{e}") 
    finally: 
        print("Execution complete")
```

Q11.  Write a function that calls another function which may raise an exception. Use try, except, and finally blocks to handle the exception and print an appropriate message.

Q13. Write a function that takes a list of strings and converts them to integers. Use try, except, and finally blocks to handle ValueError if a string cannot be converted and print an appropriate message.

```python
def convert_to_integers(lst): 
    integers = []  
    try: 
        for item in lst: 
            integers.append(int(lst)) 
    except ValueError as e: 
        print(f"Error: {e}") 
        integers = None 
    finally: 
        print("execution complete")
    return integers  
```

Q14. Write a function that uses a list comprehension to convert a list of strings to integers. Use try, except, and finally blocks within the list comprehension to handle ValueError and print an appropriate message.

```python
def convert_with_comprehension(lst): 
    try: 
        integrs = [int(item) for item in lst] 
    except ValueError as e: 
        print(f"Error: {e}") 
        integers = None 
    finally: 
        print("Execution complete") 
    return integers 
```

Q15. Write a function that attempts to write a list of strings to a file. Use try, except, and finally blocks to handle IOError and ensure the file is properly closed.

```python
def write_to_file(strings, filename): 
    try: 
        file = open(filename, 'w') 
        for string in strings: 
            file.write(string + '\n') 
    except IOError as e: 
        print(f"Error: {e}") 
    finally: 
        try: 
            file.close() 
        except NameError: 
            pass 
```
