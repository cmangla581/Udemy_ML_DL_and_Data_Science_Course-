# Understanding Exceptions

Exception handling in Python allows you too handle the errors gracefully and take the corrective actions without stopping the execution of the program. This lesson will cover the basics of the exceptions, including how to use try, except, else and finally blocks. 

# What are Exceptions ? 

Exceptions are the events that disrupt the normal flow of the program. They occur when an error is encontured during the program execution. The common exceptions include the: 

1. ZeroDivisionError : Dividing by Zero 

2. FileNotFoundError : File not Found 

3. Value Error : Invalid Value 

4. Type Error : Invalid Type 

```python
a=b  
# here, as we can see that this code gives error as a is not equal to b 
```

```python
# Exception try, except block 

try: 
    a=b 
except: 
    print("The variable has not been assigned") 

# As we can see that since a is not wqual to b, so the print statement gets printed below 
```

```python
# Now, we try the exxceptions as: 

try: 
    a=b 
except NameError as ex: 
    print(ex) 
```

As, we could see in the above code that the name error was printed stating that the name b is not defined, hence the same message is getting printed.

```python
# try, except, else block 

try: 
    num = int(input("Enter a number")) 
    result = 10/num 
except ValueError:
    print("That's not a valid number")
except ZeroDivisionError: 
    print("You cannot divide by zero") 
except Exception as ex: 
    print(ex) 
else: 
    print(f"The result is {result}") 
```

```python
# try, except, else and finally 

try: 
    num = int(input("Enter a number:"))
    result = num / 10 
except ValueError:
    print("Thsts not a valid number") 
except ZeroDividionError: 
    print("You can't divide by zero") 
else:
    print(f"The result is {result}") 
finally: 
    print("Execution Complete")  

# one thing important to note here is that the execution completss but also the execution takes place only when the result happens  
```

```python
# File handling and Exception Handling 

try: 
    file = open('example1.txt', 'r')
    content = file.read() 
    print(content) 

except FileNotFoundError: 
    print("The file does not exist")  

finally: 
    if 'file' in locals() and not file.closed(): 
        file.close() 
        print("File Close")  

# Since the file does not exist, sot the message gets printed below 
```
