# Reference Counting 
Reference Counting is the primary method Python uses to manage memory. Each object in Python maintains a count of reference pointing to it. When the reference count drops to zero, the memory occupied by the object is deallocated. 

```python
import sys 
a = [] 
## OOne reference from 'a' and one from getrefcount()  
print(sys.getrefcount(a)) 
```

```python
b = a 
print(sys.getrefcount(b)) 
```

```python
# Now, comes the process of deleting b and then making the reference as: 
del b  
print(sys.getrefcount(a))  


# Since b is deleted, so we get the reference coode as b 
```

# Garbage Collection 
Python includes a cyclic garbage collector to handle the reference cycles. Reference cycles occur when objects reference each other, preventing their reference  counts from reaching zero. 

```python
import gc 
# enable garbage collection 
gc.enable() 
```

```python
gc.disable() 
```

```python
gc.collect() 
```

```python
print(gc.get_stats()) 
```

```python
# get unreachable objects 
print(gc.garbage) 

# Here, since we do not have as such any un reachable object, so we get an empty list here 
```

# Memory Management Best Practices 

1. Use Local Variables: Local Variables have  a shorter lifespan and  are freed sooner than the global variables.  
2. Avoid circular references: Cirular references can lead to memory leaks if not  properly mmannaged.  

3. Use generators: Generators produce items one at a time and only keep one item in  a memory at a time making  them memory efficient. 

Explicitly Delete Objects: Use the del statement to delete the variables and objects explicitly. 
Profile Memory Usage: Use the memory profiling tools like the trcemalloc and memeory_profiler to identify the  memory leaks  and memory usage. 


```python
import gc 

class MyObject: 
    def __init__(self, name): 
        self.name = name  
        print(f"Object {self.name} created") 

    def __del__(self): 
        print(f"Object {self.name} deleted") 

#  Create another circular reference as:  
obj1 = MyObject("obj1") 
obj2 = MyObject("obj2") 
obj1.ref = obj1 
obj2.ref = obj2 

del obj1 
del obj2  

# Manually trigger the garbage collection 
gc.collect() 
    
```

```python
# Generators for memory efficiency 
# Generators allow you to produce the items at one time, using the memory efficiently by only keeping only one item.  

def generate_numbers(n): 
    for i in range(n): 
        yield i 

for num in generate_numbers(100000):
    print(num) 
    if num > 10: 
        break 
```

```python
# Profiling the memory usage with the tracemalloc 
import tracemalloc 

def create_list(): 
    return [i for i in range(100000)] 

def main(): 
    tracemalloc.start()

    create_list()  

    snapshot = tracemalloc.take_snapshot() 
    top_stats = snapshot.statistics('lineno') 

    print("{Top 10}") 
    for stat in top_stats[:10]: 
        print(stat) 
```

```python
main() 
```
