Q8.  Create a list of dictionaries where each dictionary represents a student with keys 'name' and 'score'. Sort the list of dictionaries by the 'score' in descending order and print the sorted list. 

```python
# The students are sorted on the basis of score as:  

students = [
    {"name" : "Alice", "score" : 88}, 
    {"name" : "Bob", "score": 95}, 
    {"name" : "Charlie", "score" : 82}, 
    {"name" : "David", "score": 90} 
]

# Sort the list by score in descending order   '
sorted_order = sorted(students, key = lambda x : x["score"], reverse = True)  

for students in sorted_order: 
    print(students)  
# This is for printing in the sorted order 
```

Q9. Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices. 

```python
# printing the transpose of the matrix as  
def transpose_matrix(matrix): 
    transposed = [] 

    for i in range(3): 
        row  = [] 
        for j in range(3): 
            row.append(matrix[j][i]) 
        transposed.append(row) 
    return transposed  

# Original Matrix 
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

# Transposiing the matrix 
result = transpose_matrix(matrix)  

# Printing the original matrix 
print("Original matrix\n") 
for row in matrix: 
    print(row)  

# prnt the transposed matrix 
print("\nTransposed Matrix\n") 
for row in result: 
    print(row) 
```

Q 10. Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists. 

```python
# The function to take a nested list and flattening into a single list. Printing the orginal and flattened lists  

def flatten_list(nested_list): 
    flat_list = [] 

    for sub_list in nested_list:  
        for item in sub_list: 
            flat_list.append(item) 
    return flat_list 

# Original Nested List 
nested_list = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
] 

flattened = flatten_list(nested_list)  

print("Original Nested List\n") 
print(nested_list) 

print("\nFlattened List\n") 
print(flattened) 
```

Q 11.  Create a list of the first 10 positive integers. Remove the elements at indices 2, 4, and 6, and insert the element '99' at index 5. Print the modified list

```python
numbers = list(range(1,11))  

# Remove the elements at the indices 2,4,6 
# Remove in reverse order to avoid index shifting  

for index in sorted([2,4,6], reverse = True): 
    numbers.pop(index)  

# Insert 99 at index 5  
numbers.insert(5,99)  

print("Modified List\n") 
print(numbers) 
```

Q13. Write a function that takes a list and returns a new list with the elements in reverse order. Print the original and reversed lists. 


```python
def reverse_list(lst): 
    reversed_list = [] 

    for i in range(len(lst) -1,-1,-1):
        reversed_list.append(lst[i]) 

    return reversed_list  

numbers = [10,20,30,40,50]  

# Reverse the list 
result = reverse_list(numbers) 

print("Original List\n") 
print(numbers) 

print("\nReversed List\n") 
print(result) 

# The line _-1,-1,-1 basically means that the rnage(start, stop, step) 
    
```

Q 14. Write a function that rotates a list by n positions. Print the original and rotated lists. 

```python
# The function for rotating the list  by n positions  
def rotate_list(lst, n): 
    n = n % len(lst) 

    # Rotate the list 
    rotated = lst[n:] + lst[:n] 
    return rotated  

# Original List  
numbers = [1,2,3,4,5] 
n =2 # Number of ppositions to rotate 

result = rotate_list(numbers, n) 

# print the original and rottted lists 
print("Original List\n") 
print(numbers) 

print("\nRotated List\n")
print(result) 
```
