                                                Introduction to Tuples

Tuples are the ordered  collections of the items that are immutable. They are similar too lists but their immutability makes them different. 

```python
# Creating a Tuple 
empty_tuple = () 
print(empty_tuple) 

print(type(empty_tuple)) 
```

In the above code, wwe can see that the empty tuple gets printed and also the type getting printed id tuple/ 

A Tuple can be converted into  a list and also a list can be converted to a tuple. We can see this by a example as: 

```python
# Conversion of list to tuple 
numbers = tuple([1,2,3,4,5,6]) 
print(numbers) 

# Here as we can see that the lsit has been converted to a tuple 
```

```python
numbers  =list((1,2,3,4,5,6)) 
print(numbers) 

# Here as we can see that the tuple gets converted to a list 
```

```python
# Printing of a mixed tuple 
mixed_tuple = (1, "Hello World", 3.24, True) 
print(mixed_tuple) 
```

```python
numbers = (1,2,3,4,5,6) 
print(numbers) 
```

```python
# Accessing the elements of a tuple  
print(numbers[0]) 

# The number at the 0th index is 1, so the number gets printed  

print(numbers[2]) 
# The number at index 0 

print(numbers[-1]) 
# The number at the last index 
```

                                           Packing and Unpacking Tuples

```python
packed_tuple = 1, "Hello World", 3.14 
print(packed_tuple) 
```

```python
# Now comes the un packing of tuple  
a,b,c = packed_tuple  
print(a) 
print(b) 
print(c) 

# Here the numbers are getting printed from a,b,c 
```

```python
# Unpacking with  * 
numbers = (1,2,3,4,5,6) 

first, *middle, last = numbers 

print(first) 
print(middle) 
print(last)  
```

As, we can see that in the above coode the numbers are getting printed as the first, last and the numbers in between the first and the last are printed together.

```python
# Nested Tuple 
nested_tuple = ((1,2,3), ("a", "b", "c"), (True, False)) 
print(nested_tuple[0]) 

print(nested_tuple[1][2]) 
```

```python
# Iterating ver the nested tuples 
for sub_tuple in nested_tuple: 
    for item in sub_tuple: 
        print(item, end = " ") 

    print() 

# Here as we can see that the all items in the nested tupple gets printed 
```
