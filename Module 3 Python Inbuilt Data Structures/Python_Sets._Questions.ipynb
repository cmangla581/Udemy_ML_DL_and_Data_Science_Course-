Q3. Create two sets: one with the first 5 positive integers and another with the first 5 even integers. Perform and print the results of union, intersection, difference, and symmetric difference operations on these sets.

```python
# creating of the 2 sets 

set1 = set(range(1,6)) 
set2 = set(num for num in range(1,11) if num%2 == 0)

print(set1)
print(set2) 
```

```python
union_set = set1.union(set2) 
print(union_set) 

intersection_set = set1.intersection(set2) 
print(intersection_set) 

difference = set1.difference(set2) 
print(difference) 

symmetric_difference = set1.symmetric_difference(set2) 
print(symmetric_difference)
```

Q12.   Create a set and remove elements from it until it is empty. Print the set after each removal. 

```python
# Creation ofna set and the removal of the elements from it, until itsempty.  

# The simple python example using pop() to remove elements until the set is empty 

s = {1,2,3,4} 

while s: 
    removed = s.pop() 
    print(f"Removed; {removed}") 
    print("Current set: ", s)
```
