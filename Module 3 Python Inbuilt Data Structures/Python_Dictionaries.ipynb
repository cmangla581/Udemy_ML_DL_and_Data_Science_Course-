                                               Introduction to Dictionaries

Dictionaries are unordered collection of items. They store the data in key value points. keys must be unique and immutable(eg. string, numbers oor tuples) while the values can be of any type. 

```python
empty_dict = {} 
print(type(empty_dict)) 

# Here the type dictionary gets printed 
```

```python
empty_dict  =dict() 
print(empty_dict) 

# A empty dictionary gets printed 
```

```python
student = {"Name" : "Krish", "age" : 32, "grade" : 24} 
print(student) 
print(type(student)) 

# Here the dictionary gets printed and hence the type dictionary gets printed 
```

```python
student = {"Name" : "Krish", "age": 32, "Name" : 24} 
print(student) 

# Here, as we can see that the name at the end is counted and the name at the starting is of no use 
```

```python
student = {"Name" : "Krish", "age": 32, "grade" : "A"} 
print(student) 
```

```python
print(student["grade"]) 
print(student["age"])  

# Here, as we can see that the grade and the age gets printed 
```

```python
# Accessing using the get() method 
print(student.get("grade")) 
print(student.get("age")) 

# Here this is the get function and hence we get the grade and the age 
```

Modfying  Dictionary Elements 
Dictionary is mutable, so you can add, update or delete elements 

```python
student["age"] = 33 
print(student) 

student["address"] = "India" 
print(student) 

# As we can see that the lists get printed 
```

```python
# Now we will be seeing the delete function  
del student["grade"] 

print(student) 

# Here as we can see that the grade key value gets deleted from the pair. 
```

```python
# Dictionary Methods 

keys = student.keys() 
print(keys) # the keys will get printed

values = student.values() 
print(values)   # the values will get printed

items = student.items() 
print(items)  # the whole dictionary will get printed 
```

Iterating over Dictionaries: 
You can use loops to iterate over dictionaries, keys and the values of the items. 

```python
# iterating over the keys 
for keys in student.keys(): 
    print(keys)  

# Here all the keys are getting printed 
```

```python
# iterate over key value pairs 
for key, value in student.items(): 
    print(f"{key}: {value}") 
```

```python
# Nested Dictionaries 

students = {"student1" : {"name": "Krish", "age": 32},
            "student2": {"name":"Peter", "age":35}} 

print(students) 
```

```python
# Access the nested dictionaries elements 
print(students["student2"]["name"]) 
print(students["student2"]["age"]) 
```

```python
# Iterating over nested dictionaries 
for student_id, student_info in students.items(): 
    print(f"{student_id}:{student_info}") 
    for key, value in student_info.items(): 
        print(f"{key}:{value}") 
```

```python
# Dictionary Comprehension 
squares = {x:x**2 for x in range(5)} 
print(squares) 
```

```python
evans = {x: x**2 for x in range(10) if x%2==0} 
print(evans)  

# Here the code is getting printed 
```

Use a diictionary to count the frequency of elements in the list 

```python
numbers =[1,2,2,3,3,3,4,4,4,4] 
frequency = {} 

for number in numbers: 
    if number in frequency: 
        frequency[number] += 1 
    else: 
        frequency[number] = 1 

print(frequency) 
```

```python
# Merge 2 dictionaries into  one 
dict1 = {"a":1, "b":2} 
dict2 =  {"b":3, "c": 4}  
merged_dict = {**dict1, **dict2} 
print(merged_dict) 
```

```python
# Using the update function 
dict1.update(dict2)  
print(dict1) 
```
