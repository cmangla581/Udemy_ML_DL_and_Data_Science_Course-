## Sets

Sets are built in the data type in Python used too store collections of unique items. They are unordered, meaning the elements do not follow a specific order and they do not allow duplicate elements.  

Sets are useful for membership tests, eliminating duplicate entries, and performing mathematial entries like the union, intersection, difference and symmetric difference. 

```python
# Creating a set 
# set is rpinted by using {} 

my_set = {1,2,3,4,5} 
print(my_set) 
print(type(my_set)) 
```

```python
my_empty_set = set() 
print(type(my_empty_set)) 

# Here as we cansee that the type of the set gets printed 
```

```python
my_empty_set = set([1,2,3,4,5,6]) 
print(my_empty_set)  

# Here the set is getting printed 
```

                                                      Basic Sets Operation

```python
my_set = {1,2,3,4,5,6} 
my_set.add(7) 

print(my_set) 
# Here as we can see that 7 gets added in the set 
```

```python
my_set.remove(3) 
print(my_set) 

# Here as we can see that 3 gets removed from the set 
```

```python
my_set.remove(10) 
# 10 is not in the list, hence it will not be removed 
```

```python
my_set.discard(11)
print(my_set)  
```

in the above code, we can see that the discard function is a bit like the remove function but there's a small difference that in case of the Discard Function, the number which is not present inset will not show any error.

# Set Operations 
Now, we will perform some of the mathematical operaations on theset 

```python
set1 = {1,2,3,4,5,6} 
set2 = {4,5,6,7,8,9} 

# Union 
union_set = set1.union(set2) 
print(union_set) 

# Here as we can see that the sets are getting printed here as union and the elements don't get repeated  

# intersection of the sets 
intersection_set  = set1.intersection(set2) 
print(intersection_set) 
# 4,5,6 are the common elements and hence they get printed as the intersection 
```

```python
set1 = {1,2,3,4,5,6} 
set2 = {4,5,6,7,8,9} 

# Difference 
set1.difference(set2) 

# the diffferemce between the sets is printed and the elements are printed here due to the individual elements as {1,2,3}
# as they are the elements oof set 1 
```

```python
# Now, we try the opposite of the above code 
set2.difference(set1) 

# 7,8,9 gets printed in the set as seperate values as they are the unique elements present in theset2 
```

```python
# Symmetric Difference 
set1.symmetric_difference(set2) 

# Here, in the case of symmetric difference, we can sse that the symmetric difference gets printed in the form of unique elements 
# fromm both the sets 
```

```python
# Sets Method 

set1 = {1,2,3} 
set2 = {3,4,5} 

# is subset function 
print(set1.issubset(set2)) 

# is superset function 
print(set1.issuperset(set2)) 
```

Here, in the above code we can see the output that, the set1 is not a subset of set2. 

Also, set1 is not a super set of set2 

```python
# Now, we try more examples of the code. 

set1 = {1,2,3,4,5} 
set2 = {3,4,5} 

print(set1.issubset(set2)) 

print(set1.issuperset(set2)) 
```

In the above coe we can see that the set1 is not a subset of set2 but its a superset of set2. 

```python
# conversion of list to the  set  
list = [1,2,2,3,4,4,5] 
set(list) 

# Here, as we can see that the list gets printed from 1,2,3,4,5 
```
