                                             Conditional Statements 

The Conditional Statements are those statements which are coded by using the if, else and elif conditions. 

```python
# if statement  

age = 20 

if age >= 18: 
    print("You are allowed to vote in elections")  

# Since the statements are true, the statement gets printed 
```

The else statement executes a block of code if the the condition in the if statement is false. 

```python
age = 10 

if age >= 18: 
    print("You are eligible for voting") 

else: 
    print("You are a minor") 

# Hence here its a minor  
```

Elif statements are those which are used to print the multiple conditions, basically more than 2 conditions. 

```python
age = 20 

if age < 13: 
    print("You are a child") 

elif age < 18: 
    print("You are a teenager") 

else: 
    print("You are an adult") 
```

Nested Conditional Statements: 
It means that we can place more if, else and elif ststements under more if, elif and else if statements to create nested else if statements. 

```python
num = int(input("Enter the number: ")) 

if num>0 : 
    print("The number is positive") 

    if num%2 == 0: 
        print("The number is even") 

    else: 
        print("The number is odd") 

else: 
    print("The number is zero or negative") 

# 34 was put as output 
```

```python
# Practical Examples of the nested if-else statements  

# Leap year question example  
year = int(input("Enter the year: ")) 

if year%4 == 0: 
    if year%100==0: 
        if year%400==0: 
            print(year, "is a leap year") 

        else: 
            print(year, "is not a leap year") 

    else: 
        print(year, "is a leap year") 

else: 
    print("Year is not a leap year") 
```

```python
# Assignment 
# Simple Calculator  

num1 = float(input("Enter the first number: ")) 
num2 = float(input("Enter the second number: ")) 
operation = input("Enter operation (+,-,*,/): ") 

# Perform the operations 
if operation == '+': 
    result = num1 + num2 

elif operation == '-': 
    result = num1 - num2 

elif operation == '*': 
    result = num1 * num2 

elif operation == '/':
    if num2 != 0: 
        result = num1 / num2 

    else: 
        print("Division by zero is not possible") 

else: 
    print("Invalid Operation") 

print("Result:", result)

# 56,8,/ 
```
