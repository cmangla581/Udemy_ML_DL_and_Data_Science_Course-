                                                       Loops

    For Loop : A For Loop is a tyee of loop which is used to iterate over a sequence oof numbers. 

```python
# Example of For Loop 
for i in range(5): 
    print(i)  

# Here as we can see that the numbers get written froom 0 to 4 
```

```python
for i in range(1,6):
    print(i)  

# Here as we want fro 1 to 5, so we write till 6 so that 6 doesn't get displayed 
```

```python
for i in range(1,10,1): 
    print(i)  

# Here, as we can see that the numbers get printed from 1 to 9 
```

```python
for i in range(1,10,2): 
    print(i)  

# Here, as we cna see that the numbers from 1 to 19 get printed with a gap of 2 in between. 
```

```python
for i in range(10,1,-1):  
    print(i) 

# Here, in this code we can see that the numbers are getting printed from 10 to 2 in reverse with a gap of 1 
```

While Loop : The While Loop is the type of loop which continues to execute as long as the condition is true. 

```python
count = 0

while count < 5: 
    print(count) 
    count = count + 1  

```

As we can see that the numbers get printed from 0 to 4 continuously without a gap. 

Loop Control Statements : 

Break : The Break statements exit the loop prematurely. 

```python
# Example of Brak Statement 

for i in range(10): 
    if i==5: 
        break 
    print(i) 
```

In the above code, we can see that the rnage we have put is equl to 10 but still in the output, the numbers are getting printed from 0 till 4 due to break statemetn. 

Continue Statement : The Continue statement is basically the type of statement which skips the current iteration and continues with the other statement. 

```python
for i in range(10): 
    if i%2 == 0: 
        continue
    print(i) 

# Here, as we can see that the numbers are getting printed from 1 to 9 with a gap of 2 
```

Pass: The pass statement is a null operation and it does nothing. 

```python
for i in range(5): 
    if i==3: 
        pass 
    print(i) 
```

```python
# Here, we can show one more example of pass statement by the pass statements 

for i in range(5): 
    if i==3:
        print("The number is: ", i)  
        pass 
    print(i) 

# Here as we cansee that the pass statement did nothing significant and the code gets executed 
```

Nested Loops:  Nested Loops basically stand for loops inside a loop.  

```python
# Example of nested loops 

for i in range(3): 
    for j in range(2): 
        print(f"i: {i} and j:{j}") 
```

In the above code, we can see that the values oof the code get printed from the 0,1,2 but the value of j from 0,1 and a total of 3x2 = 6 values are being printed.
