Q12. Write a program that calculates the factorial of a number input by the user using a while loop. 

```python
# Factorail  by while loop  
n = int(input("Enter a number: ")) 

factorial = 1 
i  =1 

while i <= n: 
    factorial = factorial*i 
    i = i+1 

print(f"Factorial of {n} is {factorial}") 
```

```python
# Factorial using the for loop 

n = int(input("Enter the number: ")) 
factorial = 1

for i in range(1, n+1):
    factorial = factorial * i 

print(f"The factorial of {n} is {factorial}") 
```

Q13. Write a program that calculates the sum of the digits of a number input by the user using a while loop. 

```python
n  =int(input("Enter the number: ")) 

sum_digits = 0  

for digit in str(n): 
    sum_digits += int(digit) 

print("Sum of digits is: ", sum_digits) 
```

```python
# using the while loop 
num = int(input("Enter a number: "))

sum_digits = 0

while num > 0:
    digit = num % 10
    sum_digits += digit
    num //= 10

print("Sum of the digits:", sum_digits)
```
