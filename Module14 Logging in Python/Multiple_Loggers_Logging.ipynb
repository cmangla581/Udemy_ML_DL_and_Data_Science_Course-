# Multiple Logging 
We can create the multiple loggers for the different parts of the application. 

```python
import logging 

# Create a  logger for module 1  
logger1 = logging.getLogger("module1") 
logger1.setLevel(logging.DEBUG) 

# Create a logger for module 2 
logger2 = logging.getLogger("module2") 
logger2.setLevel(logging.WARNING)  

# Configure the logging settings as: 
logging.basicConfig(
    level = logging.DEBUG,
    format = '%(asctime)s - %(name)s- %(levelname)s- %(message)s', 
    datefmt = '%Y - %m - %d %H:%M:%S'
) 

# It is getting executed properly 
```

```python
# Logging the messge with different loggers as: 
logger1.debug("This is the debug message for the module1") 
logger2.warning("This is a warning message for module 2") 
logger2.error("This is an error message") 
```
