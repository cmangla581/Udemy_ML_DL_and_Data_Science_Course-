# Python Logging 

Logging is a crucial aspect of any application providing a way to track the events, errors and operational information. Python's built in logging module offers a flexible framework for emitting thr log messages  from Python programs  

Here, ww eill be seeing the logging application accurately. 

```python
import logging  

# Configure the basic logging settings as: 
logging.basicConfig(level = logging.DEBUG) 

# log messages work as: 
logging.debug("This is a debug message") 
logging.info("This is an info message")  
logging.warning("This is a warning message") 
logging.error("This is an error message") 
logging.critical("This is a critical message")  
```

# Log Levels  
Python log odules has several log levels indicating the severity of events. The default levels are: 

DEBUG: Detailed information, typically of interest only when diagonising problems. 

INFO: Confirmation that the things are working as expected  

WARNING: An indication that something unexpected had happened or indicative of some problems in the near future (eg, disk space low).
          The software is still working as expected 

ERROR: A more seroius proble, due to hich the software has not been able to perform 

CRITICAL : A very serious error indicating that the program may itself be unable to  continue running. 


```python
# configuring the logging as: 
logging.basicConfig(
    level = logging.DEBUG,
    format = '%(asctime)s-%(name)s-%(levelname)s-%(message)s', 
    datefmt = '%Y - %m- %d %H:%M:%S',
    force =  True
) 

logging.debug("This is a debug message") 
logging.info("This is an info message")  
logging.warning("This is a warning message") 
logging.error("This is an error message") 
logging.critical("This is a critical message")  
```

Now, its also necessary to save the configuration file to be saved properly and hence it can be done as: 

```python
import logging 
logging.basicConfig(
    filename = 'app.log', 
    filemode = 'w',
    level = logging.DEBUG, 
    format = '%(asctime)s-%(name)s-%(levelname)s-%(message)s', 
    datefmt = '%Y-%m-%d %H;%M:%S', 
    force = True 

) 

logging.debug("This is a debug message") 
logging.info("This is an info message") 
logging.warning("This is a warning message") 
logging.error("This is an error message") 
logging.critical("This is a critical message") 
```
