```python
# Now, seperating between the independent and dependent features as: 

x = dataset.iloc[:, :-1] 
y = dataset.iloc[:, :-1] 
```

```python
from sklearn.model_selection import train_test_split 
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size =0.35, random_state = 10) 
```

```python
from sklearn.preprocessing import StandardScaler 
scaler = StandardScaler() 

scaler 
```

```python
x_train = scaler.fit_transform(x_train) 
x_test = scaler.transform(x_test) 
```

```python
x_train 
```

```python
x_test 
```

# Model Training 

```python
from sklearn.linear_model import LinearRegression 
regression = LinearRegression() 
regression 
```

```python
regression.fit(x_train, y_train) 
```

```python
regression.coef_ 
# Slope or coefficient 
```

```python
regression.intercept_  

# Here, these are the intercepts 
```

```python
# prediction for the test data 

y_pred = regression.predict(x_test) 
y_pred 
```

# Implementing the Performance Metrics 

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error 

print("MSE: ", mean_squared_error(y_test, y_pred))
print("MAE: ", mean_absolute_error(y_test, y_pred))
print("RMSE: ", np.sqrt(mean_squared_error(y_test, y_pred))) 
```

```python
from sklearn.metrics import r2_score 
score = r2_score(y_test, y_pred) 
print(score) 
```

```python
# Now, printing the adjused r2 score as: 
1 - (1 - score)**(len(y_test)-1)/(len(y_test)-x_test.shape[1]-1) 
```

# Pickling 

Python pickle module is used for the serialising and de serialising a Python object structure.  

Any object in Python can be pickled so that it can be saved on disk. What pickle does is that it “serialises” the object first before writing it to file. Pickling is a way to convert a python object (list, dict, etc.) into a character stream. The idea is that this character stream contains all the information necessary to reconstruct the object in another python script. 

```python
import pickle 
pickle.dump(regression, open('regressor.pkl', 'wb')) 
```

```python
model=pickle.load(open('regressor.pkl','rb')) 
model 
```

```python
model.predict(x_test) 
```
