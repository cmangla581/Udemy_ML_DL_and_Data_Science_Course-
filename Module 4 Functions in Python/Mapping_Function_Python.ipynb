## The Map() Function 

The map() function is the function which applies a given function to all items in an input list and returns a map object. This is particularlry useful for the transforming the data in a listcomprehensively. 

```python
def square(x): 
    return x*x 

square(10) 

# The square 100 is printed as the output 
```

```python
numbers = [1,2,3,4,5,6,7,8] 

list(map(square, numbers)) 
```

In the above code, we cansee that the square of the number gets printed in the whole list 

```python
numbers = [1,2,3,4,5,6,7,8] 

list(map(lambda x:x*x, numbers)) 

# The squares of the numbers get printed 
```

```python
# Map Multiple Iterables 

numbers1  =[1,2,3] 
numbers2  =[4,5,6] 

added_numbers = list(map(lambda x,y : x+y, numbers1, numbers2)) 

print(added_numbers) 
```

Map() to convert the numbers from the list of strings to integers 

Use map  to convert the strings to integers 

```python
str_numbers = ['1', '2', '3', '4', '5'] 

int_numbers = list(map(int,  str_numbers)) 

print(int_numbers)  

# Thestring gets converted to list 
```

```python
words = ["apple", "banana", "cherry"] 
upper_word  = list(map(str.upper, words)) 

print(upper_word) 
# Here we have capitalixed th words in the whole string 
```

```python
def get_name(person): 
    return person['name'] 

peoples = [
    {'name' : 'Krish', 'age': 32},
    {'name' : 'Jack', 'age' : 33}
]

list(map(get_name, peoples))
```
