##  Why Functions ? 

```python
# Suppose we try a code as 
num  = 24 
if num%2 == 0: 
    print("The number is even") 
else: 
    print("The number is odd") 
```

```python
def even_or_odd(num): 
    """This function finds even or odd""" 
    if num%2 == 0: 
        print("The number is even") 
    else: 
        print("The number is odd") 
```

```python
# As we can see that the function is written here 
even_or_odd(24) 
```

```python
# function with the multiple parameters 
def add(a,b): 
    return a+b 

result = add(2,4) 
print(result) 
```

Default Parmeters 

```python
# Now we take an example of a code as: 
def greet(name): 
    print(f"Hello {name} Welcome to the paradise") 

greet("Krish") 
```

In the above cde, we can see that the nam Krish gets printed  as we print the greet function.  
Now, we take a look at the greet function with a default parameter. 

```python
def greet(name = "Guest"): 
    print(f"hello {name}. Welcome to the paradise") 

greet() 

# The greet function by defaults prints the naeas guest 
```

But suppose e try by another way where we write the default value as Guest but atill put the name and the name gets printed 

```python
def greet(name = "Guest"): 
    print(f"Hello {name}. Welcome to the pardise") 

greet("Chaitanya") 

# Here the name Chaitanya gets printed even thouugh Guest is set as default 
```

```python
## Variable Length Arguments 
## Poositional and Keyword Arguments  

def print_numbers(*args): 
    for number in args: 
        print(number) #  this  is an example of positional argument 
```

```python
# here, in the above code we can see that the functions is working properly  
print(1,2,3,4,5,6,7,8,9,"Krish", "Chaitanya")  
```

```python
print(1,2,3,4,5,6,7,8,9,"Krish", "Chaitanya", sep = "\n")   
```

```python
# Now, we take the example of keyword arguments
def print_details(**kwargs): 
    for key,value in kwargs.items(): 
        print(f"{key}:{value}") 
```

```python
# Now, the above function is working, now we enter the key values 
print_details(name = "Krish", age = 32, country = "India") 
```

```python
def print_details(*args, **kwargs): 
    for val in args: 
        print(f"Positional Argument: {val}") 

    for key, value in kwargs.items(): 
        print(f"{key} : {value}") 
        
```

```python
# Now, we print ehe details as: 
print_details(1,2,3,4,"Krish", name="Krish", age = "32", country ="India") 
```
