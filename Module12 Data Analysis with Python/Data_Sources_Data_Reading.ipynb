# Reading the Data from the Different Sources 

```python
import pandas as pd 
from io import StringIO 

Data = '{"employee_name": "James", "email": "james@gmail.com", "job_profile": [{"Title1": "Team Lead", "Title2": "Sr. Developer"}]}' 
df = pd.read_json(StringIO(Data))  
df 
```

Here, as we can see that the Data Frame gets printed like this and it is formed like a table in the tabular form. 
Also, is is to note that the json is like a key value data type written above. 

```python
# Converting the data frame back to json takes place as: 
df.to_json() 
```

Here, as we can see that the data frame is getting converted back to json key value pair. 

```python
df.to_json(orient ='index') 
```

Here, as we can see that the value of the index of the table gets printed which is 0. So, basically 0 value gets pprinted as thr data frame is the 0th index. 

```python
# we can also try the records method for printing the orient method which will work as: 
df.to_json(orient = 'records')  

# Here as we can see that the tabular data frame is getting printed here in the form of reccords 
```

```python
import pandas as pd
import requests

url = "https://en.wikipedia.org/wiki/Mobile_country_code"

# Add headers (this fixes 403 error)
headers = {
    "User-Agent": "Mozilla/5.0"
}

# Get page content
response = requests.get(url, headers=headers)

# Read HTML from response text
tables = pd.read_html(response.text)

df = tables[0]

df.head()
```
