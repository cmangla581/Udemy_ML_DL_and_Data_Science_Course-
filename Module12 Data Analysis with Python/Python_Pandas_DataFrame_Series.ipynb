# Pandas  
Pandas is a powerful data manipulation library in Python which is widely used for the data analysis and also the data cleaning. It provides the 2 rimary data structures: Series and DataFrame. 

A Series is a one dimensional array like object whereas a DataFrame is a two dimensional size mutable and potentially heterogenous tabular data structure with labelled axes as the rows and columns. 

```python
import pandas as pd 
```

```python
# Series 
# Pandas Series is a one dimensional array like object that can hold any data type. It is basically similat to column in a table.  

import pandas as pd 
data = [1,2,3,4,5] 
series = pd.Series(data) 
print("Series \n", series)  

print(type(series)) 
```

In the above code, as we can see that the series are getting printed and also, the data in the series is getting printed along with the type of the data. 

```python
# Creating a series from the dictionary 
data = {'a':1, 'b':2, 'c':3} 
series_dict = pd.Series(data) 
print(series_dict)
```

Here, in the above code, we can see that the letters a, b and c become the index instead of the normal indexing from numbers. 

```python
# Printing the Series by taking the data and index seperately 
data = [10,20,30] 
index = ['a', 'b', 'c'] 

pd.Series(data, index = index)  

# Here, this is the series which are printed here 
```

```python
# Operation on DataFrame  
# Creating a DataFrame from a dictionary of list  
import pandas as pd 
data = {
    "Name":["Krish", "John", "Jack"], 
    "Age":[25,30,45], 
    "City":["Bangalore", "New York", "Florida"] 
} 

df = pd.DataFrame(data) 

print(df) 
print(type(df)) 
```

```python
# Using the numpy in the above code works as: 
import numpy as np 
np.array(df)  

# The whole data is printed in the form of Numpy 
```

# Adding of a column 
Now, suppose we have to add a column, then it takes place as: 

```python
# Adding a column as: 
df['Salary'] = [50000, 60000, 70000, 80000]
df  

# The salary column has been added 
```

```python
# Removing a column takes place as: 
df.drop('Salary', axis = 1)  

# Salary column gets removed properly and hence the rest of the data gets printed 
```

In the above code we can see that the salary column is a part of column, hence for removing it, we have to write axis = 1 which stands for column, as by default the value is set as axis = 0 which is for rows. 

```python
# Modification of the elements can be easily taken place 
# For example age column  

df['Age'] = df['Age'] + 1 
df  

# here, as we can see that the age gets upgraded 
```

```python
# Dropping the row of the 0th index takes place as 
df.drop(0)  

# 0th index row gets removed 
```
