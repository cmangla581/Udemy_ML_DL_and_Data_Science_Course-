# Handling the Missing Values 

```python
df.isnull() 

# Here, false means that there are no missing values in the data and tru means the presence of the missing values 
```

```python
df.isnull().astype(int)  
# Here 0 menas a non missing value and 0 means a non missing value 
```

```python
# Now, suppose, if we want the data in the form of the list where each is printed with true or false depending on the presence or absence
# of the missing values 

df.isnull().any(axis = 1)  
# Here false means the non missing values and true means the missing values  
# Here the axis = 1 stands for the columns and it shows the distribution column wise  
```

```python
# Getting the values that whether any missing value present in any column or not 
df.isnull().any()  
# Here, as we can see that there are no missing values present in the column 
```

Filling the missing values with the fillna function can be done as: 

```python
df.fillna(0) 
```

This is the function to fill al the missing values with the value 0, but since in this data. there are no missing values, hence the data is seen normal. 

```python
# Filling the missing values with the mean of the column 
df["Total Revenue"] = df['Total Revenue'].fillna(df['Total Revenue'].mean()) 
df 
```

The above method is basically used for printing the columns with the mean of the missing values and hence, now the missing values are being printed here. 

```python
# Prinitng the data types of the columns as: 
df.dtypes 
```

# Renaming the columns 
Here, in the code , we will be seeing to rename the column names in the whole data to perform the renaming as: 

```python
df = df.rename(columns= {"Date":"Sale Date"})  
df  

# An important note is that, we need to give the renaming in the form of key value pairs 
```

In the above code, as we can see that the data column name gets changed to sale date column name. 

```python
# Changing the data types take place as by the astype function  
df["Units Sold"] = df["Units Sold"].astype(float) 
df 
```

Here, in the above code we can see that the values in the column get converted from the integers to float. 

```python
# Filling the missing values along with the cahnging the data types takes place as: 
df["Units Sold"] = df["Units Sold"].fillna(df["Units Sold"].mean()).astype(int) 
df 
```

Here, in the above code as we can see that the float values again get converted to the integer and if there would have been missing values, then they would have converted to mean. 

```python
# Now, suppose we have to apply any function on a siingle column among the whole data .
# Then it can be done as: 

df["New Units Sold"] = df["Units Sold"].apply(lambda x:x*2) 
df.head() 
```

Here, as we can see that in the new column each value gets printed to the twice of the previous value by applying the lambda function. 

# Data Aggregation and Grouping 
Now, we will be performing the data aggragation and its grouping. 

```python
df.head(5)  
# The top 5 values of the whole data are getting printed here properly and ow the gouping will take place 
```

Here we will be grouping by among the categorial and the numerical datas and will be used to calculate the mean. 

```python
grouped_mean = df.groupby('Product Name')['Total Revenue'].mean() 
print(grouped_mean)  
```

```python
df['Product Name'].count() 
# Here there are total of 240 valuesin the whole column and the mean value of each gets printed 
```

Now, we will  be taking the 2 values into account to do the grouoing and carry the analysis of grouping the data and this time using the sum() function. 

```python
grouped_sum = df.groupby(['Product Name', 'Region'])['Units Sold'].sum() 
print(grouped_sum) 
```

So, basically based on 2 categories, we are grouping it and doing the sum of the values based on the categories. 

```python
# The above same code, we can apply by using the mean also in the data as: 
grouped_sum = df.groupby(['Product Name', 'Region'])['Units Sold'].mean() 
print(grouped_sum) 
```

# Aggregating Multiple Functions 
Now, we will be aggregating the multiple functions and it will take place as: 

```python
grouped_agg = df.groupby('Region')['Units Sold'].agg(['mean', 'sum', 'count']) 
grouped_agg 
```

Here, as we can see that the region wise analysis of the whole data is present here. 

# Merging and Joining the Data Frames 
Creating the sample data frames 

```python
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1,2,3]}) 
df2 = pd.DataFrame({'Key': ['A', 'B','D'], 'Value2': [4,5,6]}) 
```

```python
df1  
# Here df1 gets printed in the tabular form properly 
```

```python
df2  
# df2 also gets printed in the tabular form properly 
```

# Merging the 2 data frames 
Now, we will be merging the 2 dataframes by applying the codes. 

```python
pd.merge(df1, df2, on = "Key", how = "inner")  

# Here, as we can see that both these data frames ae getting merged here on the column name Key and the join taking place is inner 
```

In the above code, the inner parameter basically mean that the joining takes place on the same solumns in both, Now, we try the other joining parameters which take place as: 

```python
pd.merge(df1, df2, on = "Key", how = "outer")  
# This is because C has no value 2 and D has no value 1 
```

```python
# Trying the left parameter for joining can be done as: 
pd.merge(df1, df2, on = "Key", how = "left") 
```

In the above coode, we can see that since the df1 is present on the left, so its given more priority and the values of C get printed here. 

```python
pd.merge(df1, df2, on = "Key", how = "right")  
# Here, since d2 is on right, so its given priority 
```

```python
pd.merge(df1, df2)  
# Here by default it gets printed like inner 
```
