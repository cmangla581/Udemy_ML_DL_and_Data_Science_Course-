# Data Visualization With Seaborn 
Seaborn is a Python library based on matplotlb that provides a high level interface for drawing the attractive and informative statistical graphs. Seaborn helps in the creaing a complex visualizationsa with just a few lines of code. Here, we will cover the basics of the Seaborn including the creating the various types of plots and customizing them. 

```python
# Basic pplotting with seaborna can be done as: 
import seaborn as sns 

sns.get_dataset_names() 
# Here, we get all the datssset names as a lsit 
```

```python
# tips dataste 
tips = sns.load_dataset('tips') 
tips  

# Here, we are getting the tips dataset which is being properly printed 
```

```python
# Creating a scatter plot as: 
import matplotlib.pyplot as plt 
import seaborn as sns 

sns.scatterplot(x= 'total_bill', y = 'tip', data = tips)
plt.title("Scatter Plot of the Total Bill vs Tip") 
plt.show() 
```

In the above, code we can see that the scatter plot gets plotted and we can easily visualize the distribution of the pplot in the whole data. 

```python
# Now, we will b eplotting the line plot for the total_bill vs the size or we can say th total number of people.  
import seaborn as sns 
import matplotlib.pyplot as plt 
sns.lineplot(x = 'size', y = 'total_bill', data = tips) 
plt.title("Line Plot of the Total Bill by Size")  
plt.show()

# The line plot gets printed properly 
```

```python
# categorical Plots 
# bar Plot 

sns.barplot(x = 'day', y = 'total_bill', data = tips) 
plt.title("Bar Plot of the Total Bill day by day") 
plt.show()  

# Here as we can see that the bar plot gets printed properly 
```

```python
# Plotting of the box plot can be done as: 
import seaborn as sns 
sns.boxplot(x = "day", y = "total_bill", data = tips) 
```

```python
# Plotting the violin plot can be plotted as: 
sns.violinplot(x = 'day', y = 'total_bill', data = tips) 
```

A boxplot and the violin plot are both used to visualize the distribution of the numerical data and common groups but they emphaisze the different aspects.  A box lot provides a concise summary of the key statustics and violin plot reveals the full shape and density of the distribution plot. 

```python
# Plotting of the histogram can be done as: 
sns.histplot(tips["total_bill"], bins = 10, kde = True)  
```

```python
# If, we make kde as false then the code which we get is: 
sns.histplot(tips["total_bill"], bins = 10, kde = False) 

# Here no kde has been plotted 
```

```python
# Plotting the kde or we can say kernel density estimator plot as: 
sns.kdeplot(tips['total_bill'], shade = True) 

# here, the kde plot gets plotted properly 
```

```python
# Plotting of the pariplot 
sns.pairplot(tips)
```

A pairplot basically shows the relation of each and every feature and hence this is why a feature plot becomes mportant. 

# Plotting of the correlation heatmap 
The plotting of the correlation heatmap can be plotted as: 

```python
corr = tips[['total_bill','tip', 'size']].corr() 
corr 
```

```python
# Plotting of the heatmap 
sns.heatmap(corr, annot = True, cmap = 'coolwarm') 
```

Here, as we can see that the correlation heatmap gets plotted here and the place at 1 means that the place is heavily correlated. 

```python
import pandas as pd 
import matplotlib.pyplot as plt 

sales_df = pd.read_csv('sales_data.csv') 
sales_df.head()  

# Top 5 data get sprinted here properly 
```

```python
# Plotting the total sales by product as: 
plt.figure(figsize = (10,6)) 
sns.barplot(x = 'Product Category', y = "Total Revenue", data = sales_df, estimator = sum) 
plt.title("Total Sales by Product") 
plt.xlabel("Product") 
plt.ylabel("Total Sales") 
plt.show() 
```

Here, the bar plot of the total sales of the product wise category gets printed properly. 

```python
# Plotting the graph of total sales by region 

plt.figure(figsize = (10,6)) 
sns.barplot(x = 'Region', y = "Total Revenue", data = sales_df, estimator = sum) 
plt.title("Total Sales by Region") 
plt.xlabel("Region") 
plt.ylabel("Total Sales") 
plt.show() 
```
