                                                     Python Modules and Packages 

In Python, modules and packages help to organize and revise the code. Here's a comprehensiveguide on how to import them. 

```python
# Example of importing the math library 
import math 
# Using the square root function 

math.sqrt(16) 
```

```python
from math import sqrt, pi  

print(sqrt(16)) 
print(sqrt(4))  

print(pi)  
```

In the above code, we can see that from the math library all the calculations are  done. 

```python
# Importing the numpy array 

import numpy as np 
np.array([1,2,3,4,5,6,7,8])  

# Here,as we can see that the array gets printed using the numpy library  
```

There is one more way to import all the modules present in the math library as: 

```python
from numpy  import * 

print(sqrt(16))
print(pi)  

print(ceil(2.54))  

```

In Python, there is a vary special technique that we can create our own special packages and modules and import our packages. 

```python
# Now, we have to import the modules from the packages 

from package.maths import addition  

addition(2,3)  

# The addition takes place of the numbers 
```

```python
from package import maths 

maths.addition(4,5)  

# The addition of maths takes
```

```python
from package.subpackage.mult import multiply 

print(multiply(5,6))  

# Importin the multiply package 
```
