Q6.  Use the `os` module to create a new directory, list the contents of the current directory, and remove the newly created directory.

```python
import os 

# Create a new dictionary 
dir_name = "my_new_dictionary"  

os.mkdir(dir_name) 
print(f"Directory:'{dir_name}' created")  

# List elements of the current directory 
print("Contents of the current directory:") 
print(os.listdir(".")) 

# Remove the newly created directory 
os.rmdir(dir_name) 
print(f"Directory '{dir_name}' removed") 
```

Q9.  Use the `datetime` module to print the current date, calculate the date 100 days from today, and determine the day of the week for a given date.

```python
from datetime import date, timedelta 

# Print the current data 
today = date.today() 
print("Current date: ", today)  

# Printing the future date 
future_date = today + timedelta(days = 100) 
print("Date 100 days from today: ", future_date) 

# Determine the day of the wweek fora given week 
given_date = date(2024, 12,25) 
day_of_week = given_date.strftime("%A") 
print("Day of the week for", given_date, "is:", day_of_week) 
```

Q10. Use the `random` module to generate a list of 5 random numbers between 1 and 50 and shuffle the elements of a list.

```python
# The random module to generate a list of 5 random numbers 

random_numbers = random.sample(range(1,51), 5) 
print("Random Numbers: ", random_numbers) 

my_list = [1,2,3,4,5] 
random.shuffle(my_list) 
print("Shuffled List: ", my_list) 
```
