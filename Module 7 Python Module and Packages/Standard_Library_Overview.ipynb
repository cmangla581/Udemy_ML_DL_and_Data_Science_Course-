## Python Data Serealization 

Here, the data serealization takes place by the method of the json library. 

```python
# Data Serealization 
import json 
data = {'name' : 'Krish', 'age': 15}  

json_str = json.dumps(data) 
print(json_str) 
print(type(json_str))  

parsed_data = json.loads(json_str) 
print(parsed_data) 
print(type(parsed_data)) 
```

In the above data, as we can see that the json data type converts the data. 

```python
## Now comes the CSV file and its application

import csv

with open('example.csv', mode = 'w', newline = '') as file: 
    writer = csv.writer(file) 
    writer.writerow(['name', 'age']) 
    writer.writerow(['Krish', 32]) 
    
with open('example.csv', mode =  'r') as file: 
    reader = csv.reader(file) 
    for row in reader: 
        print(row) 
```

In the above code as we can see that the example csv file is created and the code is implemented successfully. 

```python
# Now comes the implementation of the date time library  
from datetime import datetime, timedelta 

now   = datetime.now() 
print(now) 

yesterday = now - timedelta(days = 1) 

print(yesterday) 
```

```python
# Now we will be using the time library 
import time 
print(time.time()) 

time.sleep(2) 
print(time.time()) 

# Here we get the time after the program sleeps for 2 secoonds 
```

```python
# Regular Expression Pattern (re)  
import re 

pattern = r'\d+' 
text = 'here are 123 apples' 
match  =re.search(pattern, text) 
print(match.group()) 
```

In the above code, we can see that the digits attern is recognized and the digits are printed.
