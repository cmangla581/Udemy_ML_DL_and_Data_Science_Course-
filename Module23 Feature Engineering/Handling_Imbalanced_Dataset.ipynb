# Handling Imbalanced Dataset 

Suppose, we have a dataset and we are trying to solve a classsification problem and it has categories in it. Suppose, it is a type of the Binary Classification. Now, we consiider the binary classification to understand the missing values. 

Now, suppose in the binary classification, we have the example of a dataset with 900 Yes and 100 No. Now, upon training model, it can become biased towards the category with more points. Hence, it is to be taken cae that the model does not become biased towards one cateogory only and both the categories are equally looked upon. 

It can be done mainly in 2 types of methods: 

1. Up Sampling: Here, we try to increse the samples in the lesser one. 
2. Down Sampling: Here, we try to devrese the samples in the larger one. 

```python
import numpy as np 
import pandas as pd 

np.random.seed(123)  

n_samples = 1000  
class_0_ratio = 0.9 
n_class_0 = int(n_samples * class_0_ratio) # 900 will come over here
n_class_1 = n_samples - n_class_0  # 100 data points will come over here 


```

```python
n_class_0, n_class_1 
```

```python
# Now, creating the dataframee as: 
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc = 0, scale = 1, size = n_class_0),

    'feature_2': np.random.normal(loc = 0, scale = 1, size = n_class_0), 

    'target' : [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc = 2, scale = 1, size = n_class_1),

    'feature_2': np.random.normal(loc = 2, scale = 1, size = n_class_1), 

    'target' : [1] * n_class_1
})
```

```python
# Now, concatenating the above data follows as:  

df = pd.concat([class_0, class_1]).reset_index(drop = True) 
```

```python
df.head() 
```

```python
df['target'].value_counts() 
```

```python
# Now, we will be performing the upsampling which goes on as: 
df_minority = df[df['target'] == 1] 
df_majority = df[df['target'] == 0] 

```

```python
from sklearn.utils import resample  
df_minority_upsampled = resample(df_minority, replace = True, n_samples = len(df_majority), 
         random_state = 42)
```

```python
df_minority_upsampled.shape
```

```python
df_upsampled = pd.concat([df_majority, df_minority_upsampled]) 
```

```python
df_upsampled['target'].value_counts() 
```

```python
# Nw, continuing wwith the down sampling it goes on as: 

f_minority = df[df['target'] == 1] 
df_majority = df[df['target'] == 0] 

```

```python
from sklearn.utils import resample  
df_majority_downsampled = resample(df_majority, replace = False, n_samples = len(df_minority), 
         random_state = 42)
```

```python
df_majority_downsampled.shape
```

```python
df_downsampled = pd.concat([df_minority, df_majority_downsampled]) 
```
