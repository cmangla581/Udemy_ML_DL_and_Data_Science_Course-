# Target_Guided_Ordinal_Encoding 

It is a technique which is used to to encode the categorical variables ased on the relationships with the target variables.  This encoding technique is used to when we have a  categorical vaariable with a large number of the uique categories  and we want to use this variable as a feature.  

In Target Guided Ordinal Encoding, we replaace each category in categorical variable and also with a umerical value based on the mean or the median of the target variable for that category. 

```python
import pandas as pd 

df = pd.DataFrame({
    'city' : ['New York', 'London', 'Paris', 'Tokoyo', 'New York', 'Paris'], 
    'price' : [200, 100, 300, 250, 180, 320] 

})
```

```python
df.head() 
```

```python
mean_price = df.groupby('city')['price'].mean().to_dict() 
```

```python
mean_price 
```

```python
df['city_encoded'] = df['city'].map(mean_price) 
```
