# SMOTE (Synthetic Minority Oversampling Technique) 

SMOTE (Synthetic Minority Oversampling Technique) is a technique used in the machine learnning used to address the imbalanced datasets where the minority class has significantly fewer instances than the majority class. SMOTE involves the  generation of the synthetic instances of the minority class by interpolating between the existing instances. 

Basically what happens that the upsampling technique is always not that useful as it leads to the same of the staandard deviation. Hence, we need to use the interpolating technique as because it menas the formation of the synthetic data by using the SMOTE technique. 

```python
from sklearn.datasets import make_classification 
```

```python
x,y = make_classification(n_samples = 1000, n_redundant = 0, n_features = 2, n_clusters_per_class = 1, 
                    weights = [0.90], random_state = 12) 
```

```python
import pandas as pd 
df1 = pd.DataFrame(x, columns = ['f1', 'f2']) 
df2 = pd.DataFrame(y, columns =['target']) 

final_df = pd.concat([df1, df2], axis =1) 
final_df.head() 

#  Here the dataframe has been printed 
```

```python
final_df['target'].value_counts()  

# These are the pointss which are 
```

```python
import matplotlib.pyplot as plt 
plt.scatter(final_df['f1'], final_df['f2'], c = final_df['target']) 
```

To apply SMOTE, we have to install imb learn library. 

```python
from imblearn.over_sampling import SMOTE 
```

```python
oversample = SMOTE() 
x,y = oversample.fit_resample(final_df[['f1', 'f2']], final_df['target']) 
```
