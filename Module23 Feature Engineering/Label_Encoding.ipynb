# Label Encoding 
Label Encoding and Ordinal Encoding are the 2 techniques which are used to convert the categorical data into the numerical data.  \

Label Encoding involves the assigning a unique value to each category in the variable and alsohe labels are arranged in alphabetical order or based on the frequency of categories. If we have a categorical variable with 3 possible values such as red, green and blue, we can represent it by label encoding. 

```python
import pandas as pd 
```

```python
df = pd.DataFrame({
    'color' : ['red', 'blue', 'green', 'green', 'red', 'blue']
}) 
```

```python
df.head() 
```

```python
from sklearn.preprocessing import LabelEncoder 
lbl_encoder = LabelEncoder() 
```

```python
lbl_encoder.fit_transform(df[['color']]) 
```

```python
lbl_encoder.transform([['red']]) 
```

```python
lbl_encoder.transform([['blue']]) 
```

```python
lbl_encoder.transform([['green']]) 
```

# Ordinal Encoding 
it is used to encode a  categorical data that has an intrinsic order or ranking. Each technique and each category reers to where each category is assigned a numerical value based on its position i the order.  

1. High School 
2. College 
3. Graduate 
4. Post graduate 

```python
df = pd.DataFrame({
    'size' : ['small', 'medium', 'medium', 'small', 'large']
}) 
```

```python
df.head() 
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
encoder = OrdinalEncoder(categories = [['small', 'medium', 'large']])
```

```python
encoder.fit_transform(df[['size']])
```
