Operators are the special symbols or the keyword that help into the programming and  mathematics and help and represent the actions or computations performed on one or more values known as opearnds. 

```python
# Arithmetic Operation 
a = 10 
b = 5 

add_result = a+b 
sub_result = a - b 
mul_result = a * b 
div_result = a / b  

print(add_result)  
print(sub_result)  
print(mul_result)  
print(div_result)  

# here addition takes place 
```

```python
a = 10 
b = 5 

floor_div_result = a // b 
modulus_result = a%b 
exponent_result = a**b 

print(floor_div_result) 
print(modulus_result) 
print(exponent_result) 
```

```python
# Simple calculator 

num1 = float(input("enter the first number "))
num2 = float(input("enter the second number ")) 

sum = num1 + num2 
difference = num1 - num2 
product  = num1 * num2 
quotient = num1 / num2 
remainder = num1 % num2 
exponent  = num1 ** num2 

print("Sum: ", sum) 
print("Difference: ", difference)
print("Product: ", product)
print("Quotient: ", quotient) 
print("Remainder: ", remainder) 
print("Exponent: ", exponent)

# 56 and 8 
```
