                                                      Data Types 

Datatypes are the classification of data that is used to  tell the compiler or the interpreter to on how the programmer intends to use the data. 

They determine th type of operation that can be performed on the data, the values that hte data can take and the amount of memory needed to store the data. 

Importance of Data Types in Python in Programming : 
1. Data Types ensure that the data is stored in a proper efficient way.
